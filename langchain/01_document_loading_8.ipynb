```python
def replace_recursively(text, old_substring, new_substring):
  """
  Replaces recursively occurrences of a substring with another substring.

  Args:
      text: The string to process.
      old_substring: The substring to be replaced.
      new_substring: The replacement substring.

  Returns:
      The string with all occurrences of old_substring replaced by new_substring.
  """
  while True:
    new_text = text.replace(old_substring, new_substring)
    if new_text == text:
      break
    text = new_text
  return text

# # Example usage
# text = "This string\n\nhas\n\nmultiple  lines\n\nwith  consecutive  \n\nline feeds."
# print(f'text:\n{text}\n{"-"*20}\n')

# cleaned_text = replace_recursively(text, '\n\n', '\n')
# print(f'cleaned_text:\n{cleaned_text}')
```

```python
def categorize_text(text_list, search_term):
  """
  This function categorizes a list of text strings into three categories based on a provided search term:

  Args:
      text_list: A list of text strings to be categorized.
      search_term: The term to search for within the text strings (case-insensitive).

  Returns:
      A tuple containing three lists:
          - search_list: Contains strings where the search_term is found.
          - remaining_list: Contains strings where the search_term is not found.
          - all_list: Contains all the original strings from the input list (for completeness).
  """

  search_list = []
  other_list = []
  all_list = text_list.copy()  # Create a copy to avoid modifying the original list

  for text in text_list:
    lower_text = text.lower()  # Convert to lowercase for case-insensitive search
    if search_term.lower() in lower_text:
      search_list.append(text)
    else:
      other_list.append(text)

  return search_list, other_list, all_list  # Return all three categorized lists

# Example usage
text_list = ["This is a report.pdf", "Watch this funny video!", "Some random text with doc", "fdkf", "dfjd.PDF", ".PDfxfsd"]
search_term = "pdf"  # You can change this to any search term

search_list, other_list, all_list = categorize_text(text_list, search_term)

print(f"Texts with '{search_term}':", search_list)
print("Other texts:", other_list)
print("Copy of orighinal list:", all_list)
```
